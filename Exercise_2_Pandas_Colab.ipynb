# Exercise 2: Basic Pandas Exercises

A Google Colab-ready notebook covering the Pandas examples in the exercise handout. The examples are self-contained and use current Pandas methods. Run the cells from top to bottom.

**Topics:** Series, DataFrames, column and row selection, adding and removing data, CSV files, duplicates, missing values, sorting, and grouping/aggregation.


```python
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None)
print("NumPy version:", np.__version__)
print("Pandas version:", pd.__version__)
```

## 1. Create a Series from a NumPy array

```python
values = np.array(["S", "I", "V", "A"])
series = pd.Series(values)
print(series)

# A Series can also have custom row labels.
letters = pd.Series(values, index=["first", "second", "third", "fourth"])
print("\nWith custom labels:")
print(letters)
```

## 2. Create DataFrames

```python
# An empty DataFrame
empty_df = pd.DataFrame()
print(empty_df)
```

```python
# Create a DataFrame from a list
names = ["Siva", "Rama", "Krishna"]
names_df = pd.DataFrame(names, columns=["Name"])
print(names_df)
```

```python
# Create a DataFrame from a dictionary of lists
people = {
    "ID": [1001, 1002, 1003],
    "Department": ["B.Tech", "M.Tech", "PhD"],
}
people_df = pd.DataFrame(people)
print(people_df)
```

```python
# Create a DataFrame from Series. Pandas aligns values by index.
info = {
    "one": pd.Series([1, 2, 3, 4, 5, 6], index=list("abcdef")),
    "two": pd.Series([1, 2, 3, 4, 5, 6, 7, 8], index=list("abcdefgh")),
}
series_df = pd.DataFrame(info)
print(series_df)
```

## 3. Select columns

```python
print("Column 'one':")
print(series_df["one"])

print("\nColumn 'two':")
print(series_df["two"])
```

## 4. Add columns

```python
column_df = pd.DataFrame({
    "one": pd.Series([1, 2, 3, 4, 5], index=list("abcde")),
    "two": pd.Series([1, 2, 3, 4, 5, 6], index=list("abcdef")),
})

# Add a Series as a new column; unmatched rows become NaN.
column_df["three"] = pd.Series([20, 40, 60], index=list("abc"))

# Add a column computed from existing columns.
column_df["four"] = column_df["one"] + column_df["three"]
print(column_df)
```

## 5. Delete columns

```python
delete_df = pd.DataFrame({
    "one": [1, 2, 3],
    "two": [10, 20, 30],
})
print("Original DataFrame:")
print(delete_df)

# del removes a column in place.
del delete_df["one"]
print("\nAfter deleting 'one':")
print(delete_df)

# pop removes a column and returns it.
removed_column = delete_df.pop("two")
print("\nRemoved column:")
print(removed_column)
print("DataFrame after pop:")
print(delete_df)
```

## 6. Select rows with `loc`, `iloc`, and slices

```python
row_df = pd.DataFrame(
    {
        "one": [1, 2, 3, 4, 5],
        "two": [10, 20, 30, 40, 50],
    },
    index=list("abcde"),
)
print(row_df)

# loc selects by row label.
print("\nRow with label 'b':")
print(row_df.loc["b"])

# iloc selects by integer position (zero-based).
print("\nFourth row by position:")
print(row_df.iloc[3])

# This slice selects rows at positions 2, 3, and 4.
print("\nRows in positional slice [2:5]:")
print(row_df.iloc[2:5])
```

## 7. Add rows with `concat`

`DataFrame.append()` has been removed from recent Pandas versions, so use `pd.concat()` instead.

```python
first_df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
second_df = pd.DataFrame({"A": [7, 8, 9], "B": [10, 11, 12]})
combined_df = pd.concat([first_df, second_df], ignore_index=True)
print(combined_df)
```

## 8. Delete rows

```python
rows_df = pd.DataFrame(
    [[4, 5], [6, 7], [8, 9], [10, 11]],
    columns=["x", "y"],
    index=["row0", "row1", "row2", "row3"],
)
print("Original DataFrame:")
print(rows_df)

# Drop a row using its label. Assign the result or use inplace=True.
rows_df = rows_df.drop(index="row0")
print("\nAfter dropping row0:")
print(rows_df)
```

## 9. Specify column order and access a column

```python
employee_data = {
    "Gender": ["F", "M", "M"],
    "Emp_ID": ["E01", "E02", "E03"],
    "Age": [25, 27, 25],
}
employee_df = pd.DataFrame(employee_data, columns=["Emp_ID", "Gender", "Age"])
print(employee_df)

# Bracket notation works for all column names.
print("\nEmp_ID column:")
print(employee_df["Emp_ID"])
print("First employee ID:", employee_df["Emp_ID"].iloc[0])

# Attribute access (employee_df.Emp_ID) also works for simple column names,
# but bracket notation is safer when names contain spaces or overlap with methods.
```

## 10. Read and write a CSV file

```python
# Create a small CSV in the Colab runtime, then read it with read_csv.
csv_example = pd.DataFrame({
    "Name": ["Asha", "Ben", "Chen"],
    "Score": [88, 92, 85],
})
csv_path = "sample_scores.csv"
csv_example.to_csv(csv_path, index=False)
loaded_df = pd.read_csv(csv_path)

print(loaded_df)
print("\nSummary statistics:")
print(loaded_df.describe())
print("\nFirst rows:")
print(loaded_df.head())
print("\nLast rows:")
print(loaded_df.tail())
```

## 11. Remove duplicate rows

```python
employee_info = pd.DataFrame({
    "Name": ["Siva", "Rama", "Krishna", "Siva"],
    "Age": [21, 32, 29, 21],
})
print("Before removing duplicates:")
print(employee_info)

unique_employee_info = employee_info.drop_duplicates()
print("\nAfter removing duplicates:")
print(unique_employee_info)
```

## 12. Identify missing values

```python
people_info = pd.DataFrame({
    "Person": ["Siva", "Pavan", "Gopal", "Raghu"],
    "Age": [27.0, 29.0, np.nan, 32.0],
})
print(people_info)
print("\nNon-null counts:")
print(people_info.count())
print("\nMissing-value mask:")
print(people_info.isnull())
print("\nMissing-value count per column:")
print(people_info.isnull().sum())
```

## 13. Sort rows

```python
print("Original DataFrame:")
print(people_info)

print("\nSort by index, ascending:")
print(people_info.sort_index())

print("\nSort by index, descending:")
print(people_info.sort_index(ascending=False))

print("\nSort by Age, smallest first (missing values last):")
print(people_info.sort_values("Age", na_position="last"))
```

## 14. Grouping and aggregation with the Iris dataset

The handout uses a local `iris.csv` path that is specific to one computer. This cell uses the Iris dataset bundled with scikit-learn instead, so it can run in Google Colab without uploading a separate file.

```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
iris_df = iris.frame.rename(columns={"target": "species_id"})
iris_df["species"] = iris_df["species_id"].map(
    dict(enumerate(iris.target_names))
)

print(iris_df.head())

# Aggregate only numeric columns, grouped by species.
numeric_columns = iris_df.select_dtypes(include="number").columns
mean_by_species = iris_df.groupby("species")[numeric_columns].mean()
sum_by_species = iris_df.groupby("species")[numeric_columns].sum()

print("\nMean values for each species:")
print(mean_by_species)
print("\nSum of values for each species:")
print(sum_by_species)
```
