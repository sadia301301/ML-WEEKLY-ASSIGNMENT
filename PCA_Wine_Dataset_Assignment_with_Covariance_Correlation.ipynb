## Practical 1: Covariance and Correlation Between a Pair of Attributes

**Aim:** To calculate covariance and correlation between Study_Hours and Marks using Pandas.

```python
import pandas as pd

# Create a dataset
data = {
    'Study_Hours': [2, 4, 6, 8, 10],
    'Marks': [20, 40, 50, 70, 90]
}

df = pd.DataFrame(data)

# Calculate covariance
covariance = df['Study_Hours'].cov(df['Marks'])

# Calculate correlation
correlation = df['Study_Hours'].corr(df['Marks'])

# Display results
print("Covariance:", covariance)
print("Correlation:", correlation)
```

**Expected output**

```python
Covariance: 85.0
Correlation: 0.9948497511671097
```

**Observation:** The covariance is positive, so Study_Hours and Marks tend to increase together. The correlation is close to +1, indicating a very strong positive linear relationship.

## Practical 2: Compute Covariance Matrix and Correlation Matrix

**Aim:** To compute the covariance matrix and correlation matrix for attributes A, B and C using Pandas.

```python
import pandas as pd

# Create a dataset
data = {
    'A': [2, 4, 6, 8, 10],
    'B': [1, 3, 5, 7, 9],
    'C': [5, 10, 15, 20, 25]
}

df = pd.DataFrame(data)

# Compute covariance matrix
print("Covariance Matrix:")
print(df.cov())

# Compute correlation matrix
print("\\nCorrelation Matrix:")
print(df.corr())
```

**Expected output – Covariance Matrix**

```python
A     B     C
A  10.0  10.0  25.0
B  10.0  10.0  25.0
C  25.0  25.0  62.5
```

**Expected output – Correlation Matrix**

```python
A    B    C
A  1.0  1.0  1.0
B  1.0  1.0  1.0
C  1.0  1.0  1.0
```

**Observation:** All three attributes have perfect positive correlation in this example because each attribute changes in a perfectly linear way with the others. The covariance values differ because covariance depends on the scale of the attributes.

# Part B – Principal Component Analysis (PCA) Using the Wine Dataset

The following sections use the Wine dataset and answer the five questions in the supplied PCA assignment sheet.

## 1. Import required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_wine
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

sns.set_theme(style="whitegrid")
print("Libraries imported successfully.")
```

## 2. Load and explore the Wine dataset

```python
wine = load_wine()

X = pd.DataFrame(wine.data, columns=wine.feature_names)
y = pd.Series(wine.target, name="Class")

print("First five rows:")
display(X.head())

print("Dataset shape:", X.shape)
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Wine classes:", np.unique(y))
print("\nClass names:", wine.target_names)
```

```python
print("Dataset information:")
display(X.describe())

print("Missing values in each feature:")
display(X.isnull().sum())
```

## 3. Covariance between a pair of variables

Covariance indicates how two variables vary together. A positive value indicates that the variables tend to increase together; a negative value indicates that one tends to increase as the other decreases. Its magnitude depends on the variables' measurement scales.

```python
feature_1 = "alcohol"
feature_2 = "malic_acid"

cov_value = X[feature_1].cov(X[feature_2])

print(f"Covariance between {feature_1} and {feature_2}:")
print(cov_value)
```

## 4. Correlation between a pair of variables

Correlation measures the direction and strength of a linear relationship. Its value lies between -1 and +1 and is not expressed in the original measurement units.

```python
corr_value = X[feature_1].corr(X[feature_2])

print(f"Correlation between {feature_1} and {feature_2}:")
print(corr_value)
```

```python
plt.figure(figsize=(7, 5))
sns.regplot(data=X, x=feature_1, y=feature_2)
plt.title(f"{feature_1} vs {feature_2}")
plt.show()
```

## 5. Covariance and correlation matrices

```python
cov_matrix = X.cov()
corr_matrix = X.corr()

print("Covariance matrix:")
display(cov_matrix)

print("Correlation matrix:")
display(corr_matrix)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

sns.heatmap(cov_matrix, cmap="coolwarm", ax=axes[0])
axes[0].set_title("Covariance Matrix")

sns.heatmap(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1, ax=axes[1])
axes[1].set_title("Correlation Matrix")

plt.tight_layout()
plt.show()
```

## 6. Standardize the features

The Wine features have different measurement scales. Standardization transforms each feature to have approximately zero mean and unit standard deviation. This prevents features with large numerical scales from dominating the PCA calculation solely because of their units or magnitudes.

The target class labels are kept separate. PCA is fitted only to the feature matrix `X`.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

scaled_df = pd.DataFrame(X_scaled, columns=wine.feature_names)
print("First five rows of standardized data:")
display(scaled_df.head())

print("Standardized feature means (approximately 0):")
print(np.round(X_scaled.mean(axis=0), 6))

print("Standardized feature standard deviations (approximately 1):")
print(np.round(X_scaled.std(axis=0), 6))
```

## 7. Apply PCA with two principal components

PCA creates new features called principal components. The first component captures the largest possible variance; each following component captures the largest remaining variance while being orthogonal to the earlier components.

```python
pca_2 = PCA(n_components=2)
X_pca_2 = pca_2.fit_transform(X_scaled)

pca_2_df = pd.DataFrame(X_pca_2, columns=["PC1", "PC2"])
print("Wine data after PCA (first five rows):")
display(pca_2_df.head())

print("Shape before PCA:", X_scaled.shape)
print("Shape after PCA:", X_pca_2.shape)
print("Explained variance ratio:", pca_2.explained_variance_ratio_)
print("Total variance retained:", pca_2.explained_variance_ratio_.sum())
```

## Question 1. What happens if we change the number of components?

Change `PCA(n_components=2)` to `PCA(n_components=3)`.

```python
pca_3 = PCA(n_components=3)
X_pca_3 = pca_3.fit_transform(X_scaled)

print("Shape after reducing to 3 components:", X_pca_3.shape)
print("Number of features after reduction:", X_pca_3.shape[1])
print("Number of samples:", X_pca_3.shape[0])
print("Explained variance ratio:", pca_3.explained_variance_ratio_)
print("Total variance retained:", pca_3.explained_variance_ratio_.sum())
```

**Answers – Question 1**

- **How many features are present after reduction?** Three features (PC1, PC2 and PC3).
- **Does the number of samples change?** No. The number of rows remains 178.
- **Does the total variance retained increase or decrease?** It increases or stays the same when an additional principal component is retained, because the added component contributes non-negative variance.

## Question 2. What happens if standardization is removed?

Here PCA is applied to the original, unstandardized feature values. Compare its explained variance ratio with PCA on standardized data.

```python
pca_unscaled = PCA(n_components=2)
X_pca_unscaled = pca_unscaled.fit_transform(X)

print("Explained variance ratio without standardization:")
print(pca_unscaled.explained_variance_ratio_)

print("Total variance retained without standardization:")
print(pca_unscaled.explained_variance_ratio_.sum())

print("\nExplained variance ratio with standardization:")
print(pca_2.explained_variance_ratio_)

print("Total variance retained with standardization:")
print(pca_2.explained_variance_ratio_.sum())
```

**Answers – Question 2**

- **Does the explained variance ratio change?** Yes. PCA is based on the variance of the input features, so changing the scale generally changes the components and their explained variance ratios.
- **Why might features with larger numerical scales influence PCA more strongly?** Their larger numerical variances can dominate the covariance matrix and therefore influence the principal directions more strongly.
- **Is standardization always necessary?** No. It is usually useful when features have different units or scales. It may not be needed when all features are already on comparable scales or when their original scale differences are meaningful and should be preserved.

## Question 3. What happens if we retain more components?

Try 1, 2, 3, 5 and 10 principal components. The Wine dataset has 13 original features, so each requested component count is valid.

```python
component_counts = [1, 2, 3, 5, 10]
variance_results = []

for n in component_counts:
    pca_temp = PCA(n_components=n)
    pca_temp.fit(X_scaled)
    cumulative_variance = pca_temp.explained_variance_ratio_.sum()
    variance_results.append(cumulative_variance)
    print(f"{n} component(s): cumulative explained variance = {cumulative_variance:.4f} ({cumulative_variance*100:.2f}%)")

variance_table = pd.DataFrame({
    "Number of Components": component_counts,
    "Cumulative Explained Variance": variance_results,
    "Variance Retained (%)": np.array(variance_results) * 100
})
display(variance_table)
```

```python
pca_all = PCA()
pca_all.fit(X_scaled)

cumulative_all = np.cumsum(pca_all.explained_variance_ratio_)
components_for_90 = np.argmax(cumulative_all >= 0.90) + 1

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(cumulative_all) + 1), cumulative_all, marker="o")
plt.axhline(0.90, linestyle="--", label="90% variance")
plt.axvline(components_for_90, linestyle="--",
            label=f"{components_for_90} components")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance – Wine Dataset")
plt.xticks(range(1, 14))
plt.legend()
plt.show()

print("Components needed to retain at least 90% variance:", components_for_90)
print("Variance retained:", cumulative_all[components_for_90 - 1])
```

**Answers – Question 3**

- **How does cumulative explained variance change?** It increases or stays the same as more components are retained.
- **How many components are needed to retain at least 90% variance?** Use the printed result from the cell above; it calculates this directly from the Wine dataset.
- **Is reducing to two components always the best choice?** No. Two components are useful for visualization, but more components may be needed to preserve enough information for a particular task.

## Question 4. What happens to the visualization?

Plot the first two principal components and observe the three wine classes. The class labels are used only to color the points in the plot; they are not used to calculate PCA.

```python
plt.figure(figsize=(9, 6))

for class_id, class_name in enumerate(wine.target_names):
    mask = (y == class_id)
    plt.scatter(
        X_pca_2[mask, 0],
        X_pca_2[mask, 1],
        label=class_name,
        alpha=0.8
    )

plt.xlabel("First Principal Component (PC1)")
plt.ylabel("Second Principal Component (PC2)")
plt.title("PCA Visualization of the Wine Dataset")
plt.legend(title="Wine Class")
plt.grid(True)
plt.show()
```

**Answers – Question 4**

- **Are the classes clearly separated?** The plot shows how the three classes are distributed in the first two principal-component dimensions. The degree of separation should be judged from the generated plot; some overlap may remain.
- **What information might be lost when reducing 13 features to 2?** Variance represented by the other 11 components is discarded, so some differences and patterns in the original data may no longer be visible.
- **Does PCA use the class labels to create the components?** No. PCA is unsupervised and uses only the feature values. Class labels are added afterward for visualization.

## Question 5. Does PCA select existing features?

The principal components are new variables formed as linear combinations of the original 13 Wine features.

```python
print("Original feature names:")
print(list(X.columns))

print("\nPrincipal component names:")
print(["PC1", "PC2"])

print("\nPCA component loading matrix shape:")
print(pca_2.components_.shape)

loadings = pd.DataFrame(
    pca_2.components_.T,
    index=wine.feature_names,
    columns=["PC1", "PC2"]
)
display(loadings)
```

**Answers – Question 5**

- **Are the two principal components original Wine dataset columns?** No. PC1 and PC2 are newly constructed combinations of the original features.
- **What is the difference between feature selection and feature extraction?** Feature selection chooses a subset of the existing features. Feature extraction transforms the original features into a new set of features.
- **Why is PCA considered a feature extraction technique?** PCA creates new principal components from weighted combinations of the original features rather than selecting existing columns.

## 8. Advantages and limitations of PCA

**Advantages**
- Reduces the number of features.
- Can reduce redundancy among correlated features.
- Helps visualize high-dimensional data in two or three dimensions.
- Can reduce noise when low-variance components are discarded.

**Limitations**
- Principal components can be difficult to interpret.
- Some information is lost when components are discarded.
- PCA captures linear directions of variance and may not represent nonlinear structure.
- Results can be affected by feature scaling.
